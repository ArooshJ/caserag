```python
import os
import json
import psycopg2
import datetime
from isaacus import Isaacus
from langchain.text_splitter import RecursiveCharacterTextSplitter
from psycopg2.extras import execute_values
from dotenv import load_dotenv

# Load environment variables
load_dotenv()

DB_NAME = os.environ.get("DB_NAME", "neondb")
DB_USER = os.environ.get("DB_USER", "neondb_owner")
DB_PASSWORD = os.environ.get("DB_PASSWORD", "password")
DB_HOST = os.environ.get("DB_HOST", "localhost")
DB_PORT = os.environ.get("DB_PORT", "5432")

# Categories we extract paragraphs from
TEXT_KEYS = [
    'Fact', 'Issue', "Petitioner's Argument", "Respondent's Argument",
    'Analysis of the law', 'Precedent Analysis', "Court's Reasoning", 
    'Conclusion', 'Court Disclosure', 'Uncategorized'
]

def get_db_conn():
    return psycopg2.connect(
        dbname=DB_NAME, user=DB_USER, password=DB_PASSWORD,
        host=DB_HOST, port=DB_PORT, sslmode='require'
    )

def batch_ingest_from_directory(dataset_path: str, emb_model_id: int, human_id: int):
    """
    Ingests JSON cases and generates embeddings dynamically using the Isaacus API
    based on the configuration fetched from 'ilpcsr_rag_config'.
    """
    # --- Setup Logging Directory and File ---
    log_dir = "logs"
    os.makedirs(log_dir, exist_ok=True)
    timestamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
    log_filepath = os.path.join(log_dir, f"ilpcsr_DB_Ingestion_error_Log_{timestamp}.txt")
    
    db_conn = None
    try:
        db_conn = get_db_conn()
        cursor = db_conn.cursor()

        # --- 1. Fetch Configuration ---
        print(f"Fetching configuration for emb_model_id: {emb_model_id}...")
        cursor.execute("""
            SELECT emb_model_name, chunk_size, chunk_overlap 
            FROM ilpcsr_rag_config WHERE emb_model_id = %s
        """, (emb_model_id,))
        config = cursor.fetchone()
        
        if not config:
            print(f"❌ Error: No configuration found for emb_model_id {emb_model_id}")
            return
        
        emb_model_name, chunk_size, chunk_overlap = config
        print(f"✅ Config Loaded: Model='{emb_model_name}', Chunk Size={chunk_size}, Overlap={chunk_overlap}")

        # --- 2. Initialize Models ---
        print("Initializing Isaacus API client...")
        # Automatically picks up ISAACUS_API_KEY from the environment
        isaacus_client = Isaacus() 
        
        text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=chunk_size,
            chunk_overlap=chunk_overlap
        )

        # --- 3. Pre-fetch existing data for idempotency ---
        cursor.execute("SELECT doc_id FROM ilpcsr_cases;")
        existing_cases = {row[0] for row in cursor.fetchall()}
        
        cursor.execute("SELECT doc_id FROM ilpcsr_cases_embeddings WHERE emb_model_id = %s;", (emb_model_id,))
        existing_embeddings = {row[0] for row in cursor.fetchall()}

        # --- Helper for Single Embeddings (Summary Fields) ---
        def get_single_emb(text):
            if not text or not text.strip():
                return None
            response = isaacus_client.embeddings.create(
                model=emb_model_name,
                texts=[text],
                task="retrieval/document"
            )
            return response.embeddings[0].embedding

        # --- 4. Process Directory ---
        print(f"Scanning directory '{dataset_path}'...")
        
        for filename in os.listdir(dataset_path):
            if not filename.endswith('.json'):
                continue
            
            # --- WORKLOAD DISTRIBUTION CHECK ---
            try:
                file_id_str = filename.replace('.json', '')
                file_id = int(file_id_str)
            except ValueError:
                print(f"Skipping {filename}: Cannot extract a numeric ID from filename.")
                continue
            
            if file_id % 4 != human_id:
                print(f"Skipping {filename}: Modulo result ({file_id % 4}) does not match HUMAN_ID ({human_id}).")
                continue

            # ==========================================
            # INNER TRY-EXCEPT FOR PER-FILE PROCESSING
            # ==========================================
            try:
                filepath = os.path.join(dataset_path, filename)
                with open(filepath, 'r', encoding='utf-8') as f:
                    case_data = json.load(f)

                case_id = case_data.get("caseID")
                summary = case_data.get("Summary")

                if not case_id or not summary:
                    print(f"-> Skipping '{filename}' (Missing ID or Summary).")
                    continue

                print(f"\n=> Processing '{filename}' (caseID: {case_id})...")

                # --- PHASE 1: TEXT INGESTION (WRITE-ONCE) ---
                if case_id not in existing_cases:
                    print("   [Text] Case not in DB. Inserting text, paragraphs, and chunks...")
                    source_url = f"https://indiankanoon.org/doc/{case_id}/"
                    
                    # Insert Base Case
                    cursor.execute("""
                        INSERT INTO ilpcsr_cases (
                            doc_id, source_url, title, court, case_date, relevant_statutes, relevant_precedents
                        ) VALUES (%s, %s, %s, %s, %s, %s, %s);
                    """, (
                        case_id, source_url, case_data.get('title'), case_data.get('court'),
                        case_data.get('date') or None, 
                        case_data.get('relevant_statutes', []), case_data.get('relevant_precedents', [])
                    ))

                    # Insert Paragraphs & Chunks
                    total_chunks = 0
                    for category in TEXT_KEYS:
                        paragraphs_list = case_data.get(category, [])
                        for paragraph_text in paragraphs_list:
                            if not paragraph_text.strip():
                                continue

                            cursor.execute("""
                                INSERT INTO ilpcsr_paragraphs (doc_id, metadata_label, full_text)
                                VALUES (%s, %s, %s) RETURNING paragraph_id;
                            """, (case_id, category, paragraph_text))
                            paragraph_id = cursor.fetchone()[0]

                            child_chunks = text_splitter.split_text(paragraph_text)
                            for chunk_text in child_chunks:
                                cursor.execute("""
                                    INSERT INTO ilpcsr_content_chunks (paragraph_id, chunk_text)
                                    VALUES (%s, %s);
                                """, (paragraph_id, chunk_text))
                                total_chunks += 1
                    
                    print(f"   [Text] Inserted {total_chunks} chunks.")
                    existing_cases.add(case_id) 
                else:
                    print("   [Text] Case text already exists. Skipping text ingestion.")

                # --- PHASE 2: EMBEDDING INGESTION (WRITE-MANY) ---
                if case_id in existing_embeddings:
                    print(f"   [Embed] Embeddings for Model {emb_model_id} already exist. Skipping.")
                    continue
                
                print(f"   [Embed] Generating and inserting embeddings via Isaacus API...")
                
                # 1. Embed Case Summary (Individual calls)
                facts_text = summary.get('key_facts', '')
                issues_text = "; ".join(summary.get('legal_issues', []))
                statutes_text = "; ".join(summary.get('key_statutes_cited', []))
                precedents_text = "; ".join(summary.get('key_precedents_cited', []))
                holding_text = summary.get('holding', '')

                cursor.execute("""
                    INSERT INTO ilpcsr_cases_embeddings (
                        doc_id, emb_model_id, facts_emb, issues_emb, statutes_emb, precedents_emb, holding_emb
                    ) VALUES (%s, %s, %s, %s, %s, %s, %s);
                """, (
                    case_id, emb_model_id,
                    get_single_emb(facts_text),
                    get_single_emb(issues_text),
                    get_single_emb(statutes_text),
                    get_single_emb(precedents_text),
                    get_single_emb(holding_text)
                ))

                # 2. Fetch all chunk texts from the DB for this case
                cursor.execute("""
                    SELECT c.chunk_id, c.chunk_text 
                    FROM ilpcsr_content_chunks c
                    JOIN ilpcsr_paragraphs p ON c.paragraph_id = p.paragraph_id
                    WHERE p.doc_id = %s;
                """, (case_id,))
                
                chunks_to_embed = cursor.fetchall()
                
                # 3. Embed and insert chunk embeddings (Batched for API limits)
                if chunks_to_embed:
                    # Isaacus limits 128 texts per request; batching by 100 for safety
                    BATCH_SIZE = 100
                    for i in range(0, len(chunks_to_embed), BATCH_SIZE):
                        batch = chunks_to_embed[i:i + BATCH_SIZE]
                        batch_ids = [c[0] for c in batch]
                        batch_texts = [c[1] for c in batch]
                        
                        response = isaacus_client.embeddings.create(
                            model=emb_model_name,
                            texts=batch_texts,
                            task="retrieval/document"
                        )
                        
                        # Sort embeddings by index to maintain exact alignment with batch_ids
                        batch_embeddings = [item.embedding for item in sorted(response.embeddings, key=lambda x: x.index)]
                        
                        # Defensive check: catch silent misalignment if the API ever
                        # returns a different count than requested (partial failures, etc.)
                        if len(batch_embeddings) != len(batch_ids):
                            raise ValueError(
                                f"Embedding count mismatch: sent {len(batch_ids)} texts, "
                                f"got {len(batch_embeddings)} embeddings back "
                                f"(batch starting at chunk index {i})."
                            )
                        
                        # Single round trip for the whole batch instead of one per row
                        execute_values(
                            cursor,
                            """
                            INSERT INTO ilpcsr_chunk_embeddings (chunk_id, emb_model_id, embedding) VALUES %s
                            """,
                            [(chunk_id, emb_model_id, embedding) for chunk_id, embedding in zip(batch_ids, batch_embeddings)]
                        )
                
                print(f"   [Embed] Inserted {len(chunks_to_embed)} chunk embeddings.")
                
                db_conn.commit()

            except Exception as file_err:
                # Rollback transaction for this file, log the API/DB error, and continue
                if db_conn:
                    db_conn.rollback()
                
                error_msg = f"Failed to process {filename}: {str(file_err)}"
                print(f"   ❌ {error_msg}. Skipping to next file...")
                
                with open(log_filepath, 'a', encoding='utf-8') as log_f:
                    log_f.write(f"[{datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S')}] {filename} - ERROR: {str(file_err)}\n")
                
                continue

        print("\nBatch ingestion complete! ✅")
        print(f"Check '{log_filepath}' for any skipped files/errors.")

    except psycopg2.Error as e:
        print(f"Critical Database error during setup: {e}")
        if db_conn:
            db_conn.rollback()
    except Exception as e:
        print(f"A critical unexpected error occurred: {e}")
        if db_conn:
            db_conn.rollback()
    finally:
        if db_conn:
            db_conn.close()

def build_vector_indexes(emb_model_id: int):
    """
    Dynamically builds Partial HNSW indexes for a specific embedding model 
    after the data has been ingested.
    """
    db_conn = None
    try:
        db_conn = get_db_conn()
        cursor = db_conn.cursor()

        print(f"\nFetching dimensions to build indexes for emb_model_id: {emb_model_id}...")
        cursor.execute("SELECT dimensions FROM ilpcsr_rag_config WHERE emb_model_id = %s", (emb_model_id,))
        result = cursor.fetchone()
        
        if not result:
            print("❌ Error: Configuration not found. Cannot build indexes.")
            return
            
        dimensions = result[0]
        print(f"✅ Dimensions loaded: {dimensions}. Building HNSW indexes (this may take a minute)...")

        case_vectors = ['facts_emb', 'issues_emb', 'statutes_emb', 'precedents_emb', 'holding_emb']
        
        for col in case_vectors:
            index_name = f"idx_cases_{col}_model_{emb_model_id}"
            print(f"   -> Building index: {index_name}")
            cursor.execute(f"""
                CREATE INDEX IF NOT EXISTS {index_name} 
                ON ilpcsr_cases_embeddings 
                USING hnsw (({col}::vector({dimensions})) vector_cosine_ops) 
                WHERE emb_model_id = {emb_model_id};
            """)

        chunk_index_name = f"idx_chunks_emb_model_{emb_model_id}"
        print(f"   -> Building index: {chunk_index_name}")
        cursor.execute(f"""
            CREATE INDEX IF NOT EXISTS {chunk_index_name} 
            ON ilpcsr_chunk_embeddings 
            USING hnsw ((embedding::vector({dimensions})) vector_cosine_ops) 
            WHERE emb_model_id = {emb_model_id};
        """)

        db_conn.commit()
        print("✅ All partial vector indexes built successfully!")

    except psycopg2.Error as e:
        print(f"Database error while building indexes: {e}")
        if db_conn:
            db_conn.rollback()
    finally:
        if db_conn:
            db_conn.close()

# --- RUN SCRIPT ---
if __name__ == "__main__":
    import sys
    
    human_id_str = os.environ.get("HUMAN_ID")
    if human_id_str is None or human_id_str.strip() == "":
        print("❌ CRITICAL ERROR: 'HUMAN_ID' is not set in your .env file.")
        sys.exit(1)
        
    try:
        HUMAN_ID = int(human_id_str)
    except ValueError:
        print(f"❌ CRITICAL ERROR: 'HUMAN_ID' in your .env must be a number.")
        sys.exit(1)
    
    DATASET_DIR = os.path.join("..", "il-pcsr_dataset", "corpus")
    
    # Update this to match the DB row you just inserted
    TARGET_EMB_MODEL_ID = 3 
    
    if not os.path.isdir(DATASET_DIR):
        print(f"Error: Dataset directory '{DATASET_DIR}' not found.")
    else:
        print(f"Starting pipeline for HUMAN_ID {HUMAN_ID}...")
        batch_ingest_from_directory(DATASET_DIR, TARGET_EMB_MODEL_ID, HUMAN_ID)
        build_vector_indexes(TARGET_EMB_MODEL_ID)
```
